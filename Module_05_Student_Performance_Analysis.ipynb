```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('StudentsPerformance.csv')

print("Dataset Loaded Successfully!")
```

```python
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
df.describe()
```

```python
average_marks = df[['math score', 'reading score', 'writing score']].mean()

average_marks.plot(kind='bar', figsize=(8,5))

plt.title('Average Marks of Students')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['math score'], bins=10, kde=True)

plt.title('Distribution of Math Scores')
plt.xlabel('Math Marks')
plt.ylabel('Number of Students')

plt.show()
```

```python
print("DATA ANALYSIS FINDINGS")

print("Average Math Score:", df['math score'].mean())
print("Average Reading Score:", df['reading score'].mean())
print("Average Writing Score:", df['writing score'].mean())

print("Total Students:", len(df))
```
